## DATA CLEANING
Steps: (1) work on a copy, (2) convert text columns to numbers, (3) fix units, (4) handle placeholder values, (5) missing values, (6) locations, (7) dates, (8) outliers, (9) save.

**Note on missing values:** `Size` and `Bathrooms` are only *flagged* here, not filled. Filling them with a median now would use information from the future test set. The imputation is done later inside the modelling pipeline, after the train/test split.

```python
# Work on a copy so the raw data stays untouched
clean = df.copy().reset_index(drop=True)
print('Rows at start:', len(clean))
```

### 1. Price
`Price` is text like `KSh 22,500,000`. Some listings say `Price not communicated`. Without a price there is no target, so those rows cannot be used.

```python
print('Rows with no price:', (clean['Price'] == 'Price not communicated').sum())

clean['Price'] = pd.to_numeric(clean['Price'].str.replace(r'[^\d.]', '', regex=True), errors='coerce')
clean = clean.dropna(subset=['Price']).reset_index(drop=True)
print('Rows after dropping unpriced listings:', len(clean))
clean['Price'].describe().apply(lambda x: f'{x:,.0f}')
```

### 2. Bedrooms and Bathrooms
Strip the words `Bedrooms` / `Bathrooms` and store as numbers.

```python
for col in ['Bedrooms', 'Bathrooms']:
    clean[col] = pd.to_numeric(clean[col].str.extract(r'(\d+)')[0])   # missing bathrooms stay NaN

clean[['Bedrooms', 'Bathrooms']].describe()
```

### 3. Size: mixed units
`Size` mixes **sqm**, **m²**, **sq ft** and **acre** (including fractions like `1/2 acre`). Everything is converted to square metres.

```python
import numpy as np

def to_sqm(text):
    if pd.isna(text):
        return np.nan
    t = text.lower().replace(',', '').strip()
    m = re.match(r'^(\d+)\s*/\s*(\d+)', t)          # fractions such as 1/2 acre
    if m:
        value = int(m.group(1)) / int(m.group(2))
    else:
        m = re.match(r'^([\d.]+)', t)
        if not m:
            return np.nan
        value = float(m.group(1))
    if 'acre' in t:
        return value * 4046.86
    if 'sq ft' in t or 'sqft' in t:
        return value / 10.7639
    return value                                       # sqm or m²

import re
clean['Size_sqm'] = clean['Size'].apply(to_sqm)
print(clean['Size'].str.extract(r'([a-zA-Z²]+(?: ft)?)')[0].value_counts(), '\n')
clean['Size_sqm'].describe()
```

### 4. Placeholder and impossible sizes
`20 sqm` is the most common size in the data. A house of 20 sqm is not realistic, so it looks like a default value on the listing site. Anything smaller is also impossible for a house with bedrooms. At the other end, two listings claim `2500 acre` and `5000 acre`, which are typing errors. All of these are set to missing rather than trusted.

```python
print('Rows with exactly 20 sqm:', (clean['Size_sqm'] == 20).sum())
print('Rows below 20 sqm:', (clean['Size_sqm'] < 20).sum())
print('Rows above 100 acres:', (clean['Size_sqm'] > 100 * 4046.86).sum())

clean.loc[(clean['Size_sqm'] <= 20) | (clean['Size_sqm'] > 100 * 4046.86), 'Size_sqm'] = np.nan
clean['Size_missing'] = clean['Size_sqm'].isna().astype(int)
print('Size missing now:', clean['Size_missing'].sum(), f"({clean['Size_missing'].mean():.0%})")
```

### 5. Remaining missing values
`Amenities` and `Surroundings`: a blank means nothing was listed, so they become an empty string (no statistics are used, so there is no leakage).

```python
clean['Amenities'] = clean['Amenities'].fillna('')
clean['Surroundings'] = clean['Surroundings'].fillna('')
clean['Bathrooms_missing'] = clean['Bathrooms'].isna().astype(int)

clean[['Bathrooms', 'Size_sqm', 'Amenities', 'Surroundings']].isnull().sum()
```

### 6. Location
Locations look like `Sub-area, Area` (e.g. `Runda, Westlands`) or just `Kitengela`. Two columns are created so modelling can use the broad area as well as the specific estate.

```python
clean['Location'] = clean['Location'].str.strip()
parts = clean['Location'].str.rsplit(',', n=1, expand=True)
clean['Area'] = parts[1].fillna(parts[0]).str.strip()
clean['Sub_Location'] = parts[0].str.strip()

print('Unique locations:', clean['Location'].nunique(), '| unique areas:', clean['Area'].nunique())
clean['Area'].value_counts().head(15)
```

### 7. Dates
`Created At` exists only for the 800 original listings.

```python
clean['Created At'] = pd.to_datetime(clean['Created At'], format='%d-%b-%y', errors='coerce')
clean['Created At'].describe()
```

### 8. Sanity checks and outliers
Outliers are **flagged, not deleted**. Expensive homes in Runda or Karen are real, so removing them would hurt the model where it matters.

```python
# Logical check: more bathrooms than bedrooms + 2 is unusual
clean['Bath_gt_bed_plus2'] = (clean['Bathrooms'] > clean['Bedrooms'] + 2).astype(int)
print('Listings with bathrooms > bedrooms + 2:', clean['Bath_gt_bed_plus2'].sum())

# Price outliers on the log scale (IQR rule)
lp = np.log(clean['Price'])
q1, q3 = lp.quantile([.25, .75])
iqr = q3 - q1
clean['Price_outlier'] = ((lp < q1 - 1.5 * iqr) | (lp > q3 + 1.5 * iqr)).astype(int)
print('Price outliers (log IQR):', clean['Price_outlier'].sum())

# Size outliers
ls = np.log(clean['Size_sqm'])
q1, q3 = ls.quantile([.25, .75])
iqr = q3 - q1
clean['Size_outlier'] = ((ls < q1 - 1.5 * iqr) | (ls > q3 + 1.5 * iqr)).astype(int)
print('Size outliers (log IQR):', clean['Size_outlier'].sum())

clean.loc[clean['Price_outlier'] == 1, ['Title', 'Price', 'Size_sqm', 'Record Type']].sort_values('Price').head(10)
```

### 9. Duplicates after cleaning
Now that columns are numeric, check again on the key columns. Original listings with identical specs but different URLs can be different units in the same development, so they are reported but not dropped.

```python
key = ['Property Type', 'Price', 'Location', 'Bedrooms', 'Bathrooms', 'Size_sqm', 'Amenities']
dup = clean.duplicated(subset=key, keep=False)
print(pd.crosstab(dup, clean['Record Type'], rownames=['duplicate on key columns']))
```

### 10. Final check and save

```python
final_cols = ['Record Type', 'Property Type', 'Price', 'Location', 'Area', 'Sub_Location',
              'Bedrooms', 'Bathrooms', 'Bathrooms_missing', 'Size_sqm', 'Size_missing',
              'Amenities', 'Surroundings', 'Created At',
              'Price_outlier', 'Size_outlier', 'Bath_gt_bed_plus2']
clean = clean[final_cols]
print(clean.shape)
display(clean.dtypes.to_frame('dtype').join(clean.isnull().sum().to_frame('missing')))
clean.to_csv('clean_data.csv', index=False)
clean.head()
```
